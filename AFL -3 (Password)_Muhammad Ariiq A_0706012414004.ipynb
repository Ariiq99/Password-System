```python
import hashlib
```

```python
# Target Hashes
target_A = "0110b96270248f746ecca06f1ce09746"
target_B = "8d4c531eb4b0f54df72aa6839abbeaec"
target_C = "b20f8b0405b88e2b0b50eb3356d34ba7"
target_D = "0110b96270248f746ecca06f1ce09746"
```

# 1. Plain Brute-force

```python
for i in range(1000000):
    pin = f"{i:06d}"
    if hashlib.md5(pin.encode()).hexdigest() == target_A:
        print(f"[#1 SUCCESS] PIN Plain Target A: {pin}")
        break
```

# 2. Salt Brute-force

```python
for salt in range(10):
    for i in range(1000000):
        pin = f"{i:06d}"
        if hashlib.md5(f"{salt}{pin}".encode()).hexdigest() == target_B:
            print(f"[#2 SUCCESS] Salt Target B: {salt}, PIN: {pin}")
            break
```

# 3. Salt + KDF 3x Brute-force

```python
for salt in range(10):
    for i in range(1000000):
        pin = f"{i:06d}"
        h1 = hashlib.md5(f"{salt}{pin}".encode()).hexdigest()
        h2 = hashlib.md5(h1.encode()).hexdigest()
        h3 = hashlib.md5(h2.encode()).hexdigest()
        if h3 == target_C:
            print(f"[#3 SUCCESS] Salt Target C: {salt}, PIN: {pin}")
            break
```

# 4. Dictionary Attack

```python
with open("pinlist.txt", "r") as f:
    for line in f:
        pin = line.strip()
        if hashlib.md5(pin.encode()).hexdigest() == target_D:
            print(f"[#4 SUCCESS] Wordlist Target D PIN Found: {pin}")
            break
```
